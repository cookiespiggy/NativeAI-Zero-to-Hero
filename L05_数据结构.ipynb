## 概念讲解：数据结构 = 收拾房间的不同柜子

数据乱放就像衣服堆地上。不同的「柜子」适合不同东西：

| 结构 | 柜子类型 | 特点 | 例子 |
|------|----------|------|------|
| `list` 列表 | 一排抽屉 | 有顺序、可改 | `["苹果","香蕉"]` |
| `dict` 字典 | 带标签的格子 | 用 key 查 value | `{"name":"小明"}` |
| `set` 集合 | 去重写字板 | 不重复 | `{"a","b"}` |
| `tuple` 元组 | 封死的抽屉 | 不可改 | `(1, 2)` |

## 第一步：列表 list —— 有序的一排盒子

```python
fruits = ["苹果", "香蕉", "橙子"]
print("第一个：", fruits[0])       # 索引从 0 开始！
fruits.append("西瓜")             # 追加
print("现在：", fruits)
print("长度：", len(fruits))
```

## 第二步：字典 dict —— 用名字查内容

```python
student = {"name": "小明", "age": 18, "city": "北京"}
print("姓名：", student["name"])
student["age"] = 19              # 修改
student["score"] = 95           # 新增
print("更新后：", student)
```

## 第三步：集合 set —— 自动去重

```python
votes = ["A", "B", "A", "C", "B", "A"]
unique = set(votes)              # 自动去掉重复
print("原始投票：", votes)
print("不重复的候选人：", unique, "共", len(unique), "位")
```

# 🎯 AHA 顿悟单元格：你的「智能通讯录 / 投票分析器」

运行下面代码。你会看到一个**能自动统计投票、找出冠军、并生成排行榜**的小程序。
试着往 `ballots` 里加几个名字，再运行——排行榜会立刻重算。

> 真实世界的 AI 数据处理，99% 就是「把数据装进 list/dict，再循环统计」。你已经摸到门道了。

```python
# ===== 运行我！往 ballots 里加名字试试 =====
ballots = ["猫", "狗", "猫", "兔", "狗", "猫", "狗", "猫", "鸟", "猫", "狗"]

# 用 dict 当计数器：键=候选，值=票数
count = {}
for b in ballots:
    count[b] = count.get(b, 0) + 1

# 按票数从高到低排序
ranking = sorted(count.items(), key=lambda x: x[1], reverse=True)

print("  🗳️  投票结果实时统计：")
print("  " + "=" * 28)
medal = ["🥇", "🥈", "🥉"]
for i, (name, c) in enumerate(ranking):
    m = medal[i] if i < 3 else "  "
    bar = "█" * c
    print(f"  {m} {name:<3} {bar} {c}票")
print("  " + "=" * 28)
print(f"  🏆 冠军是「{ranking[0][0]}」，共 {ranking[0][1]} 票！")
```

**AHA 机制**：实时投票排行榜，改数据即重算，直观展示「结构+循环」的工程价值。
