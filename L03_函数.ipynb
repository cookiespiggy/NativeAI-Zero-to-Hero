## 概念讲解：函数 = 一台「输入→输出」的小机器

现实里，微波炉是一台机器：**放入食物（输入）→ 按下启动 → 出来热饭（输出）**。
编程里的**函数（function）** 一模一样：

```
def 机器名(输入原料):
    加工过程...
    return 成品
```

定义一次，之后想用多少次都行。这就是「**不要重复自己（DRY）**」的工程铁律起点。

## 第一步：定义你的第一台小机器

````python
# 定义一台「平方机」：输入一个数，返回它的平方
def square(x):
    return x * x

print(square(3))    # 输入 3  → 输出 9
print(square(10))   # 输入 10 → 输出 100
````

## 第二步：机器可以有多个输入，也能自己做事

````python
# 一台「欢迎机」：输入名字和次数，打印对应问候
def greet(name, times):
    for i in range(times):           # range 生成 0,1,2... 序列，L04 细讲
        print(f"你好，{name}！({i + 1})")

greet("小红", 3)
````

## 第三步：用函数消灭重复（真实价值）

不用函数，算 5 个人的 BMI 要写 5 遍公式；用函数，只要一句。

````python
def bmi(weight_kg, height_m):
    """根据体重(kg)和身高(m)计算 BMI 指数"""
    return weight_kg / (height_m ** 2)

people = [("A", 70, 1.75), ("B", 55, 1.60), ("C", 90, 1.80)]
for name, w, h in people:
    print(f"  {name}: BMI = {bmi(w, h):.1f}")
````

# 🎯 AHA 顿悟单元格：你的「AI 健康顾问」

运行下面代码。你会得到一个**输入身高体重，立刻给出健康建议的 AI 顾问**。
试着改 `weight` 和 `height`：它会像真顾问一样，自动判断你偏瘦/正常/偏胖。

> 你刚才写的 `bmi_advice` 函数，本质上就是现代 AI 应用里「接收输入→推理→返回建议」的迷你版！

````python
# ===== 运行我！改 weight / height 看看 =====
def bmi_advice(weight_kg, height_m):
    """输入身高体重，返回一句健康建议（带智能判断）"""
    score = weight_kg / (height_m ** 2)
    if score < 18.5:
        return score, "偏瘦，注意营养均衡 🍚"
    elif score < 24:
        return score, "非常标准，继续保持 💪"
    elif score < 28:
        return score, "偏胖，建议多动一动 🏃"
    else:
        return score, "肥胖，建议咨询专业医生 🩺"

weight = 70.0      # 👈 改这里（公斤）
height = 1.75      # 👈 改这里（米）

score, advice = bmi_advice(weight, height)
print("  🤖 AI 健康顾问为您分析：")
print(f"     身高 {height}m  体重 {weight}kg")
print(f"     BMI = {score:.1f}")
print(f"     💡 建议：{advice}")
````
