# L04 · 循环与条件：让机器不知疲倦

**学习目标**
- 掌握 `for` 循环：让同一段代码重复执行
- 掌握 `if / elif / else` 条件：让程序「看情况办事」
- 理解「嵌套」：循环里套条件，条件里套循环

**前置依赖**：L01-L03（变量、类型、函数）  
**预计时长**：40 分钟  
**技术栈**：纯 Python 标准库（`time` 用于动画）

---

## 概念讲解：循环是「不知疲倦的工人」，条件是「路口的红绿灯」

- **循环 `for`**：你让它做 100 次，它绝不会喊累。人类抄 100 遍会崩溃，机器秒完成。
- **条件 `if`**：到路口看红绿灯，红灯停绿灯行。程序据此走不同分支。

两者结合，就能让机器处理「成千上万条数据，每条还要分别判断」这种人类最怕的活。

## 第一步：`for` 循环 —— 重复的艺术

```python
# range(5) 生成 0,1,2,3,4，循环体执行 5 次
for i in range(5):
    print(f"这是第 {i + 1} 次执行")
```

## 第二步：`if` 条件 —— 看情况办事

```python
scores = [95, 60, 78, 45, 88]
for s in scores:
    if s >= 85:
        print(f"  {s} 分 → 优秀 🌟")
    elif s >= 60:
        print(f"  {s} 分 → 及格")
    else:
        print(f"  {s} 分 → 不及格，加油 💪")
```

## 第三步：计数与累加（循环的真实用处）

算总和、数个数，是 AI 数据处理里最基础的操作。

```python
nums = [3, 7, 2, 9, 4]
total = 0
for n in nums:
    total += n          # 等价于 total = total + n
print("总和 =", total, " 平均 =", total / len(nums))
```

# 🎯 AHA 顿悟单元格：会「呼吸」的星空 ⭐

运行下面代码。你会看到一片**会一闪一闪、自动生成的 ASCII 星空动画**——
完全由 `for` 循环 + `if` 条件驱动，没有任何手工绘制。

> 想想看：你只是写了「规则」，星空却「活」了过来。这就是循环与条件组合的魔力，
> 也是一切 AI 批量生成内容的底层逻辑（生成、遍历、筛选）。

```python
# ===== 运行我！一片由规则生成的星空会动起来 =====
import random, time, os

WIDTH, HEIGHT, FRAMES = 40, 12, 30

def draw_sky():
    """用循环+条件，随机生成一帧星空"""
    lines = []
    for y in range(HEIGHT):
        row = ""
        for x in range(WIDTH):
            # 随机决定这一格是星星还是空格（条件判断）
            if random.random() < 0.18:
                row += random.choice("✦✧·")
            else:
                row += " "
        lines.append(row)
    return "\n".join(lines)

print("  ✨ 由 for 循环 + if 条件实时生成的星空（按星号密度呼吸）✨")
for f in range(FRAMES):
    os.system("clear" if os.name == "posix" else "cls")  # 清屏
    print(draw_sky())
    time.sleep(0.15)   # 每帧停顿 0.15 秒，形成动画
print("  🌌 动画结束。你刚刚指挥计算机画了", FRAMES, "帧星空！")
```
