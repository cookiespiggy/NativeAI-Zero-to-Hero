## 概念讲解：DataFrame = 会算账的 Excel

现实中的数据（用户表、销售记录）都是「表格」。
**pandas 的 DataFrame** 就是程序里的智能表格：你告诉它「帮我找出所有消费超过 100 的人」，它一秒算完。

AI 工程师 80% 的时间在「洗数据、查数据」，pandas 就是那把主刀。

## 第一步：造一张表并查看

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "姓名": ["小明", "小红", "小刚", "小丽", "小华"],
    "城市": ["北京", "上海", "北京", "广州", "上海"],
    "消费": [120, 80, 200, 60, 150],
})
print(df)
```

## 第二步：筛选 —— 回答一个问题

```python
# 找出消费超过 100 的「高价值用户」
high = df[df["消费"] > 100]
print("高价值用户：\n", high)
print("\n他们的平均消费：", high["消费"].mean())
```

## 第三步：分组统计 —— 城市维度对比

```python
by_city = df.groupby("城市")["消费"].mean()
print("各城市平均消费：\n", by_city)
```

# 🎯 AHA 顿悟单元格：你的「AI 用户行为分析仪」

运行下面代码。计算机会**造出 500 个虚拟用户**，然后像资深分析师一样：
自动算出「哪个城市最赚钱」「谁是高价值用户 Top3」「消费随年龄怎么变」。
改 `N` 造更多人，结论会实时刷新。

> 你刚刚做的「分组、筛选、排序、统计」正是所有推荐系统/增长分析/A-B 测试的基础动作。

```python
# ===== 运行我！改 N 造更多用户 =====
import pandas as pd
import numpy as np

np.random.seed(42)
N = 500
cities = np.random.choice(["北京", "上海", "广州", "深圳"], N)
ages = np.random.randint(18, 60, N)
# 消费与年龄轻度相关 + 噪声，模拟真实世界
spend = (ages * 3 + np.random.normal(50, 40, N)).round().clip(10, 1000)

users = pd.DataFrame({"城市": cities, "年龄": ages, "消费": spend})

print("  📊 已生成", N, "名用户，自动分析如下：")
print("  " + "-" * 40)
print("  各城市平均消费：")
print(users.groupby("城市")["消费"].mean().round(1).to_string())
print("\n  高价值用户 Top3：")
print(users.sort_values("消费", ascending=False).head(3).to_string(index=False))
print("\n  消费 vs 年龄相关性：", round(users["消费"].corr(users["年龄"]), 3))
print("  " + "-" * 40)
print("  🤖 一个 AI 分析师的日常工作，你几行代码就复现了！")
```
