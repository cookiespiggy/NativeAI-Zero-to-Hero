## 概念讲解：计算机是「死板」的会计

现实世界里，「3 个苹果」和「苹果」这两个词你都懂。
但计算机是个**死板的会计**：它必须精确知道手里的东西是「数字」还是「文字」，
否则它会报错。

这种「东西的种类」在编程里叫**数据类型（type）**。
本课我们认识 4 种最常用类型：

| 类型 | 名字 | 例子 | 用途 |
|------|------|------|------|
| `int` | 整数 | `18` | 个数、年龄 |
| `float` | 小数 | `3.14` | 温度、价格 |
| `str` | 文字 | `"你好"` | 名字、句子 |
| `bool` | 真假 | `True` / `False` | 判断、开关 |

## 第一步：用 `type()` 看穿一个东西的本质

```python
a = 18
b = 3.14
c = "你好"
d = True

print("a 是", type(a), "值：", a)
print("b 是", type(b), "值： ", b)
print("c 是", type(c), "值： ", c)
print("d 是", type(d), "值： ", d)
```

## 第二步：类型不同，能做的事也不同

数字可以加减，文字可以拼接（`+` 把两段文字连起来）。

```python
print(1 + 1)             # 数字相加 → 2
print("你好" + "世界")    # 文字拼接 → 你好世界

# 注意：数字和文字不能直接相加，会报错！
# print("年龄" + 18)     # ❌ 取消注释运行会报错，这就是「类型不匹配」
```

## 第三步：在类型之间「搬家」（类型转换）

用 `str()` 把数字变文字、`int()` 把文字变整数。这是新手必会的技巧。

```python
age = 18
print("我今年 " + str(age) + " 岁")   # 先转成文字才能拼接

num_text = "42"
print(num_text + " + 8 =", int(num_text) + 8)  # 转成整数才能算数
```

# 🎯 AHA 顿悟单元格：你的「智能小账本」

运行下面代码，你会得到一个**会自己算总价、会判断你买得起买不起的小账本**。
试着修改 `budget`（预算）和商品数量，感受「类型」如何支撑起一个真实有用的程序。

> 一行行数据背后，全是本课讲的 `int / float / str / bool` 在默默分工。

```python
# ===== 运行我！可以改预算和数量 =====
budget = 100.0                      # float：你的预算（元）
items = [("苹果", 5, 3.0), ("牛奶", 2, 8.5), ("面包", 3, 4.0)]

total = 0.0                         # float：累计总价
print("店员：欢迎光临，这是您的购物清单 👇")
print("-" * 36)
for name, qty, price in items:      # 这里先感受循环，L04 会细讲
    subtotal = qty * price
    total += subtotal
    print(f"  {name:<4} x{qty}  ¥{price:.1f}  = ¥{subtotal:.1f}")
print("-" * 36)
print(f"  合计：¥{total:.1f}   预算：¥{budget:.1f}")

can_afford = total <= budget        # bool：买得起吗？
if can_afford:
    print("  ✅ 预算充足，结账成功！")
else:
    print(f"  ⚠️ 超预算 ¥{total - budget:.1f}，请放下几件 🛒")
```
