## 概念讲解：图是数据的「照片」

看 1000 个数字你的大脑会宕机；但一张图，眼睛 0.1 秒就懂。
**可视化** 就是把抽象数字变成「形状」：趋势、异常、簇群，一目了然。
所有 AI 论文、Dashboard、监控大屏，背后都是这堂课的技能。

## 第一步：折线图 —— 看趋势

```python
import matplotlib.pyplot as plt
import numpy as np

days = np.arange(1, 8)
sales = [12, 15, 13, 20, 25, 30, 28]
plt.plot(days, sales, marker="o", color="#2ca02c")
plt.title("一周销售额"); plt.xlabel("星期"); plt.ylabel("销售额")
plt.show()
```

## 第二步：柱状图 —— 比大小

```python
cities = ["北京", "上海", "广州", "深圳"]
avg = [260, 310, 180, 290]
plt.bar(cities, avg, color="#ff7f0e")
plt.title("各城市平均消费"); plt.ylabel("元")
plt.show()
```

## 第三步：散点图 —— 找关系

```python
np.random.seed(0)
x = np.random.rand(50) * 10
y = 2 * x + np.random.normal(0, 3, 50)   # 带噪声的线性关系
plt.scatter(x, y, color="#9467bd", alpha=0.7)
plt.title("年龄 vs 消费（疑似正相关）")
plt.xlabel("年龄"); plt.ylabel("消费")
plt.show()
```

# 🎯 AHA 顿悟单元格：你的「数据侦探仪表盘」

运行下面代码。计算机会**造一段「用户增长数据」，并一键画出 4 张图**：
折线（增长趋势）、柱状（渠道对比）、散点（付费意愿）、饼图（构成）。
你会像看监控大屏一样，「一眼读懂」一家公司。

> 这就是 AI 工程师每天看的 Dashboard 的缩小版。你会了这手，离「看懂业务」只差真实数据。

```python
# ===== 运行我！看 4 张图组成的数据大屏 =====
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(7)

months = np.arange(1, 13)
users = 1000 * (1.15 ** months) + np.random.normal(0, 200, 12)  # 指数增长+噪声
channels = ["短视频", "搜索", "社交", "直邮"]
ch_val = [4200, 3100, 1800, 900]
age = np.random.randint(18, 60, 200)
pay = age * 4 + np.random.normal(0, 30, 200)
pie = [55, 25, 20]

fig, ax = plt.subplots(2, 2, figsize=(11, 7))
ax[0,0].plot(months, users, marker="o", color="#2ca02c")
ax[0,0].set_title("📈 用户增长趋势"); ax[0,0].set_xlabel("月份")
ax[0,1].bar(channels, ch_val, color="#ff7f0e")
ax[0,1].set_title("📊 获客渠道对比")
ax[1,0].scatter(age, pay, color="#9467bd", alpha=0.6)
ax[1,0].set_title("💡 年龄 vs 付费意愿"); ax[1,0].set_xlabel("年龄")
ax[1,1].pie(pie, labels=["免费","会员","企业"], autopct="%d%%", colors=["#1f77b4","#ff7f0e","#2ca02c"])
ax[1,1].set_title("🥧 用户构成")
plt.tight_layout()
plt.show()
print("  🖥️  数据大屏已生成：趋势/对比/关系/构成，四图合一！")
```
