## 概念讲解：类 = 乐高模具，对象 = 用模具造出的积木

现实里先有「手机设计图」（类 class），再按图造出「你手里的这台手机」（对象 object）。

- **属性**：手机的颜色、电量 —— 对象的「数据」
- **方法**：手机能打电话、充电 —— 对象的「行为」

用面向对象，你能把「一群有数据、会做事的东西」管理得清清楚楚——这是写大程序的地基。

## 第一步：定义一个「类」

```python
class Pet:
    def __init__(self, name, species):
        self.name = name            # 属性：名字
        self.species = species      # 属性：品种
        self.energy = 100          # 属性：体力

    def play(self):                # 方法：玩耍
        self.energy -= 20
        return f"{self.name} 玩了一会儿，体力剩 {self.energy}"

    def feed(self):                # 方法：喂食
        self.energy += 30
        return f"{self.name} 吃饱了，体力升到 {self.energy}"
```

## 第二步：用类造出真实对象

```python
cat = Pet("咪咪", "猫")
dog = Pet("旺财", "狗")
print(cat.play())
print(dog.feed())
print(f"{cat.name} 是 {cat.species}，{dog.name} 是 {dog.species}")
```

## 第三步：一群对象一起工作

把对象放进列表，循环调度——这就是「多智能体」系统的雏形。

```python
pets = [Pet("咪咪", "猫"), Pet("旺财", "狗"), Pet("吱吱", "仓鼠")]
for p in pets:
    print(p.play())
```

# 🎯 AHA 顿悟单元格：你的「AI 宠物乐园」

运行下面代码。你会看到**三只宠物各自有体力、会自己玩耍/吃饭/睡觉，甚至会对你说话**。
你可以用 `pets[0].feed()` 单独喂某只。这就是「对象 = 有状态、会互动的小个体」。

> 现代 AI Agent（智能体）系统，本质就是「很多会自己决策、互相通信的对象」。你今天写的宠物，和 SOTA 论文里的 Agent，是同一种思维方式！

```python
# ===== 运行我！也可以课后单独执行 pets[0].feed() =====
class Pet:
    def __init__(self, name, species, mood="开心"):
        self.name = name
        self.species = species
        self.energy = 80
        self.mood = mood

    def play(self):
        self.energy = max(0, self.energy - 15)
        self.mood = "兴奋" if self.energy > 30 else "疲惫"
        return f"🐾 {self.name}({self.species}) 玩耍后体力 {self.energy}，心情：{self.mood}"

    def feed(self):
        self.energy = min(100, self.energy + 25)
        self.mood = "满足"
        return f"🍖 {self.name} 进食后体力 {self.energy}，心情：{self.mood}"

    def status(self):
        return f"  {self.name}: 体力 {self.energy}/100  心情 {self.mood}"

# 造一个宠物乐园
pets = [Pet("咪咪", "猫"), Pet("旺财", "狗"), Pet("吱吱", "仓鼠")]
print("  🏞️  欢迎来到 AI 宠物乐园！")
print("  " + "-" * 30)
for p in pets:
    print(p.status())
print("  " + "-" * 30)
print("  🎮 模拟互动...")
print("  " + pets[0].play())
print("  " + pets[1].feed())
print("  " + pets[2].play())
print("  " + "-" * 30)
for p in pets:
    print(p.status())
print("  ✨ 每个对象都活了过来——这就是面向对象的魅力！")
```
