## 概念讲解：向量化 = 让一万个数同时做运算

Python 的列表一次只能处理一个元素（要写循环）。
而 **numpy 数组** 像一个「并行工厂」：**一条指令，整列数同时算完**。

这不仅是快——现代 AI（神经网络）的本质，就是海量数字的矩阵运算。
numpy 就是这一切的入口。

## 第一步：数组与向量化加法（对比循环）

```python
import numpy as np

# 普通列表：要加得写循环
a = [1, 2, 3, 4]
b = [10, 20, 30, 40]
c = [a[i] + b[i] for i in range(len(a))]

# numpy：整列同时加
arr_a = np.array(a)
arr_b = np.array(b)
arr_c = arr_a + arr_b

print("列表循环结果：", c)
print("numpy 向量化：", arr_c)
```

## 第二步：随机模拟 —— AI 世界的「抽签」

```python
# 一次生成 5 个 0~1 之间的随机数（概率直觉的雏形）
rolls = np.random.rand(5)
print("随机抽签：", rolls)

# 统计：AI 评测里天天在算的均值/最值
print("平均值：", rolls.mean())
print("最大值：", rolls.max())
```

## 第三步：矩阵 —— 数据的二维表格

```python
M = np.array([[1, 2], [3, 4]])
v = np.array([10, 20])
print("矩阵 M：\n", M)
print("向量 v：", v)
print("M 每行加 v（广播）：\n", M + v)   # 广播：自动对齐维度
```

# 🎯 AHA 顿悟单元格：蒙特卡洛掷骰子 · 看见概率的「形状」

运行下面代码。计算机会**瞬间掷 100 万次骰子，并画出分布直方图**。
你会「看见」概率长什么样——均匀分布平如地板。改 `n` 或换成两个骰子相加，
你会看到著名的「钟形曲线」自己浮现出来。

> 这就是 AI 的核心思维：用海量数字实验代替空想。numpy 让你「一念之间，百万次实验」。
> 深度学习里每一次训练，都是这样的矩阵运算在 GPU 上跑 billions 次。

```python
# ===== 运行我！可改 n 或 dice=2 看不同分布 =====
import numpy as np
import matplotlib.pyplot as plt

n = 1_000_000        # 掷骰子次数（一百万！）
dice = 1            # 骰子个数：改成 2 看钟形曲线

rolls = np.sum(np.random.randint(1, 7, size=(n, dice)), axis=1)

plt.figure(figsize=(8, 4))
plt.hist(rolls, bins=range(1, dice * 6 + 3), color="#4C72B0",
         edgecolor="white", alpha=0.85)
plt.title(f"掷 {dice} 颗骰子 × {n:,} 次的分布（numpy 一瞬间算完）")
plt.xlabel("点数"); plt.ylabel("出现次数")
plt.show()
print(f"  🎲 一百万次实验已完成：均值 {rolls.mean():.2f}，最常出现 {np.bincount(rolls).argmax()}")
```
