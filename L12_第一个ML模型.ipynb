## 概念讲解：机器学习 = 用例子喂出「规律提取器」

上节课我们手写了一个「学直线」的循环。现实中，大牛们把这些循环封装好了，
就是 **scikit-learn**。你只要：

1. 准备「带答案的例子」（训练集）
2. `model.fit(例子, 答案)` —— 学习
3. `model.predict(新数据)` —— 预测

这就是所有 AI 应用的两步式骨架。

## 第一步：造一个「两类花朵」数据集

```python
from sklearn.datasets import make_classification
import numpy as np

X, y = make_classification(n_samples=300, n_features=2, n_classes=2,
                          n_informative=2, n_redundant=0, random_state=0)
print("特征形状：", X.shape, " 标签示例：", y[:10])
print("含义：每朵花有 2 个测量值，标签 0/1 表示两个品种")
```

## 第二步：训练 / 测试划分 + 训练模型

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=0)
model = DecisionTreeClassifier(max_depth=3, random_state=0)
model.fit(X_tr, y_tr)                  # 学习！
acc = model.score(X_te, y_te)          # 在没见过的数据上打分
print(f"模型在测试集准确率：{acc*100:.1f}%")
```

## 第三步：预测一朵新花

```python
new_flower = [[2.0, -1.5]]              # 两个测量值
pred = model.predict(new_flower)
print(f"这朵新花被判定为品种：{pred[0]}（0 或 1）")
```

# 🎯 AHA 顿悟单元格：你的「AI 花朵分类器」可视化

运行下面代码。你会看到一张**决策边界图**：机器自动在两堆花之间画出分界线，
测试点被标成正确/错误。然后输入你自己的新花坐标，AI 当场告诉你它是哪个品种。

> 你刚刚完成了「数据采集→训练→评估→预测」的完整 ML 流程——这正是任何 AI 产品的缩影，
> 从垃圾邮件识别到自动驾驶，骨架完全相同。

```python
# ===== 运行我！看 AI 画出的决策边界 =====
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

X, y = make_classification(n_samples=300, n_features=2, n_classes=2,
                          n_informative=2, n_redundant=0, random_state=0)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=0)
model = DecisionTreeClassifier(max_depth=4, random_state=0).fit(X_tr, y_tr)

# 画决策边界
xx, yy = np.meshgrid(np.linspace(X[:,0].min(), X[:,0].max(), 200),
                     np.linspace(X[:,1].min(), X[:,1].max(), 200))
Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(7, 5.5))
plt.contourf(xx, yy, Z, alpha=0.25, cmap=plt.cm.coolwarm)
plt.scatter(X_te[:,0], X_te[:,1], c=y_te, edgecolors="k", cmap=plt.cm.coolwarm)
acc = model.score(X_te, y_te)
plt.title(f"AI 花朵分类器 · 测试准确率 {acc*100:.1f}%")
plt.xlabel("特征1"); plt.ylabel("特征2")
plt.show()

# 现场预测一朵你指定的新花
new = np.array([[1.5, -1.0]])
print(f"  🌸 你输入的新花坐标 {new[0]} → AI 判定为品种 {model.predict(new)[0]}")
```
