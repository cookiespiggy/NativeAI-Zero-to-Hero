## 概念讲解：计算机是个「超听话的笨助手」

想象你有一个助手，**永远不累、永不忘记、但完全不会自己思考**。
你只能用它听得懂的「咒语」（代码）下命令，它就会 100% 照做。

Python 就是其中一种最人性化的「咒语语言」。
而 **Jupyter Notebook（你正在看的这个界面）** 是一个可以「边写咒语边看结果」的魔法书。

👉 绿色方框叫「代码单元格」，按 `Shift + Enter` 就能让计算机执行它。

## 第一步：让计算机开口说话

`print(...)` 的意思是「把括号里的内容显示出来」。

```python
# print 会把内容打印到屏幕上
print("你好，我是你的计算机助手！")
print("从今天起，你下的每一个命令，我都会立刻照做。")
```

## 第二步：计算机也会做数学

计算机最擅长算数。把算式直接写进去，它就会算出答案。

```python
print(1 + 1)          # 加法
print(10 * 6)         # 乘法：一周有 10 天？不，这里是举例
print(2 ** 10)        # ** 表示乘方：2 的 10 次方等于多少？
```

## 第三步：让计算机「记住」东西

用 `=` 可以把一个值「存」进一个名字里，这个过程叫**赋值**。
被存起来的东西叫「变量」，你可以把它理解成贴了标签的盒子。

```python
name = "小明"          # 把文字存进名为 name 的盒子
age = 18               # 把数字存进名为 age 的盒子
print("我叫", name, "，今年", age, "岁。")

# 盒子里的东西可以拿来用
print(name, "明年就", age + 1, "岁了。")
```

# 🎯 AHA 顿悟单元格：会「个性化打招呼」的魔法

运行下面这段代码。试着先把上面 `name` 改成你自己的名字，再运行一次——
你会发现：**你只是改了一个词，计算机就为你量身定制了回应**。

> 这就是编程最神奇的地方：你写的是「规则」，计算机执行的是「为你而变的结果」。

```python
# ===== 运行我！可以改上面的 name 试试 =====
import datetime

name = "小明"                     # 👈 改成你的名字
now = datetime.datetime.now()
hour = now.hour

# 根据当前时间，给出不同的问候（这就是「智能」的雏形：条件判断）
if hour < 6:
    greeting = "夜深了，注意休息哦"
elif hour < 12:
    greeting = "早上好，新的一天开始啦"
elif hour < 18:
    greeting = "下午好，元气满满"
else:
    greeting = "晚上好，今天辛苦了"

print("=" * 40)
print(f"  {greeting}，{name}！")
print(f"  现在的时间是 {now.strftime('%Y-%m-%d %H:%M')}")
print(f"  这是你用代码创造的第一个智能问候 ✨")
print("=" * 40)
```
