# L11 · AI 第一性原理：从猜想到学习

## 概念讲解：AI = 一个不断纠错的「猜谜机器」

很多人以为 AI 靠魔法。其实最朴素的 AI 只有三步循环：

1. **猜**（用当前参数做预测）
2. **比**（和正确答案比，算出误差）
3. **改**（朝「减少误差」的方向微调参数）

重复成千上万次，机器就「学会」了。这就是**机器学习的第一性原理**。
深度学习、ChatGPT，本质都是这个循环被放大了亿万倍。

## 第一步：我们想学什么？一条直线的规律

真实世界：$y = 3x + 2$（但机器一开始不知道，它要自己猜出 3 和 2）。

```python
import numpy as np
np.random.seed(1)
x = np.linspace(0, 10, 50)
y_true = 3 * x + 2 + np.random.normal(0, 1.5, 50)   # 真实规律+噪声
print("机器看到的只是 (x, y) 一堆点，不知道公式是 3x+2")
```

## 第二步：让机器「猜→比错→改」

```python
w, b = 0.0, 0.0        # 初始猜测：权重0，偏置0
lr = 0.01             # 学习率：每次改多大
losses = []
for step in range(300):
    y_pred = w * x + b
    loss = ((y_pred - y_true) ** 2).mean()   # 误差（均方误差）
    losses.append(loss)
    # 梯度：误差对 w、b 的导数（朝减少误差的方向）
    grad_w = 2 * ((y_pred - y_true) * x).mean()
    grad_b = 2 * (y_pred - y_true).mean()
    w -= lr * grad_w
    b -= lr * grad_b
print(f"学完后：y ≈ {w:.2f}x + {b:.2f}  （真相是 3.00x + 2.00）")
```

## 第三步：看误差如何「跌落」

```python
import matplotlib.pyplot as plt
plt.plot(losses, color="#d62728")
plt.title("学习曲线：误差随训练步数下降")
plt.xlabel("训练步"); plt.ylabel("误差")
plt.show()
```

# 🎯 AHA 顿悟单元格：亲眼看见机器「学会」一条直线

运行下面代码。你会看到一幅**动态图**：一开始机器猜的线乱七八糟（红色），
随着训练步数增加，它一点点「纠正」自己，最终完美贴合蓝色数据点。
下方还实时滚动显示当前的 `w、b、误差`。

> 你刚刚用 20 行代码，重现了深度学习训练的核心循环。ChatGPT 也是这样「猜→改」出来的，
> 只是它的公式是千亿参数的巨网，数据是互联网全文。原理，和你看到的这条线一模一样。

```python
# ===== 运行我！看机器从乱猜到学会 =====
import numpy as np, matplotlib.pyplot as plt, time, os
np.random.seed(1)
x = np.linspace(0, 10, 50)
y_true = 3 * x + 2 + np.random.normal(0, 1.5, 50)

w, b, lr = 0.0, 0.0, 0.01
plt.ion(); fig, ax = plt.subplots(figsize=(7, 4.5))
for step in range(1, 301):
    y_pred = w * x + b
    loss = ((y_pred - y_true) ** 2).mean()
    grad_w = 2 * ((y_pred - y_true) * x).mean()
    grad_b = 2 * (y_pred - y_true).mean()
    w -= lr * grad_w; b -= lr * grad_b
    if step % 20 == 0 or step == 1:
        ax.clear()
        ax.scatter(x, y_true, s=15, c="#1f77b4", label="真实数据")
        ax.plot(x, w*x + b, c="#d62728", lw=2.5, label=f"机器猜测 y={w:.2f}x+{b:.2f}")
        ax.set_title(f"训练第 {step} 步  误差={loss:.2f}")
        ax.legend(); fig.canvas.draw(); time.sleep(0.05)
print(f"  🎓 机器学会了：y ≈ {w:.2f}x + {b:.2f}（真相 3.00x + 2.00）")
plt.ioff(); plt.show()
```
