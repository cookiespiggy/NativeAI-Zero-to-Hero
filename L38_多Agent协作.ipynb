## 项目蓝图：一个团队胜过单打独斗

复杂任务一个人做容易漏。多 Agent 系统让**专精不同角色的 AI 协作**：

```
研究员(搜集素材) → 作家(组织成文) → 审查员(挑错把关) → 成品
```
本课用面向对象把每个角色封装成独立 Agent，串成流水线。

## 第一步：用类封装三个角色（来自 L06）

````python
import random

class Researcher:
    def work(self, topic):
        return [f"关于『{topic}』的素材A", f"关于『{topic}』的素材B", f"数据：{random.randint(10,99)}%"]

class Writer:
    def work(self, materials):
        return "报告草稿：" + "；".join(materials)

class Reviewer:
    def work(self, draft):
        issues = ["缺少引用" for _ in range(random.randint(0,2))]
        if issues:
            return draft + "（需修订：" + ",".join(issues) + "）"
        return draft + "（审查通过✅）"

print("✅ 三个 Agent 角色封装完成：研究员 / 作家 / 审查员")
````

## 第二步：编排成协作流水线

````python
def collaborate(topic):
    r, w, v = Researcher(), Writer(), Reviewer()
    log = []
    m = r.work(topic); log.append(("研究员", m))
    d = w.work(m);     log.append(("作家", d))
    final = v.work(d); log.append(("审查员", final))
    return log, final

log, final = collaborate("AI 对教育的影响")
for role, out in log:
    print(f"  [{role}] {out}")
````

研究员搜集、作家成稿、审查员把关，最终产出一份带质量标记的成品。改 `topics` 批量跑多篇。

````python
# ===== 运行我！看多 Agent 协作产出 =====
topics = ["远程办公的利弊", "新能源车的未来", "短视频对青少年的影响"]
print("  🤝 多 Agent 协作系统启动（研究员→作家→审查员）\n")
for t in topics:
    log, final = collaborate(t)
    print(f"  📌 主题：{t}")
    for role, out in log:
        print(f"     {role}: {out}")
    print()
print("  🚀 三个 AI 角色成功协作，自动产出多份带质检的报告！")
````
